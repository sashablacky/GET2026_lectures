```python
class DataSource:
    def __init__(self, source_name, file_path):
        self.source_name = source_name
        self.file_path = file_path

    def display(self):
        print("Source: ", self.source_name)
        print("path: ", self.file_path)

sourcec1 = DataSource("SalesCSV", "data/sales.csv")
sourcec1.display()
source2 = DataSource("wergnon", "jobjqgew")
source2.display()
```

```python
class DatabaseSource:
    def __init__(self, host, port=5432, database="warehouse"):
        self.host = host
        self.port = port
        self.database = database

    def display(self):
        print(self.host, self.port, self.database)

db1 = DatabaseSource("localhost")
db2 = DatabaseSource("db-server", 5433, "analytics")

db1.display()
db2.display()
```

```python
class DataSource:
    source_count = 0

    def __init__(self, name):
        self.name = name
        DataSource.source_count += 1

s1 = DataSource("Sales")
s2 = DataSource("Customers")
s3 = DataSource("Products")

print(DataSource.source_count)
print(s1.source_count)
```

```python
class Pipeline:
    def __init__(self, name):
        self.name = name
        self.logs = []

    def add_log(self, message):
        self.logs.append(message)

p1 = Pipeline("Sales")
p2 = Pipeline("Customers")
p1.add_log("Sales started")
print(p1.logs)
print(p2.logs)
```

```python
class DataBatch:
    def __init__(self, records):
        self.records = records

    def record_count(self):
        return len(self.records)

    def add_record(self, record):
        self.records.append(record)

batch = DataBatch([101, 102, 103])
print(batch.record_count())
batch.add_record(104)
print(batch.record_count())
```

```python
class DataSource:
    source_count = 0

    def __init__(self, name):
        self.name = name
        DataSource.source_count += 1

    @classmethod
    def get_source_count(cls):
        return cls.source_count

s1 = DataSource("Sales")
s2 = DataSource("Customers")

print(DataSource.get_source_count())
```

```python
class DataValidator:
    @staticmethod
    def is_valid_port(port):
        return isinstance(port, int) and 1 <= port <= 65535

print(DataValidator.is_valid_port(5432))
print(DataValidator.is_valid_port(70000))
```
