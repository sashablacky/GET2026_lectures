```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

def all_prime(n):
    primes = []
    for num in range(2, n + 1):
        is_prime = True
        for prime in primes:
            if prime * prime > num:
                break
            if num % prime == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
            yield num

def find_armstrong_numbers(n):
    for num in range(n + 1):
        digits = [int(d) for d in str(num)]
        if sum(d ** len(digits) for d in digits) == num:
            yield num

def find_perfect_numbers(n):
    for num in range(2, n + 1):
        divisors_sum = sum(i for i in range(1, num) if num % i == 0)
        if divisors_sum == num:
            yield num


def display_strong():
    pass
```
