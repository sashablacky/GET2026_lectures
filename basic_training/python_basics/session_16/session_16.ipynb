```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 13500, 18000, 22000, 25000]
})

plt.figure(figsize=(8, 4))
plt.plot(
    sales_df["Month"],
    sales_df["Sales"],
    marker="o",
    linestyle="-",
    linewidth=2,
    label="Sales"
)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
sales_df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Product_A": [100, 120, 140, 135, 160],
    "Product_B": [90, 110, 115, 145, 155]
})

plt.figure(figsize=(8, 4))

plt.plot(sales_df["Month"], sales_df["Product_A"],
         marker="o", label="Product A")
plt.plot(sales_df["Month"], sales_df["Product_B"],
         marker="s", label="Product B")

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
data = {
    "Student": [
        "Anu", "Ravi", "Meera", "Arun", "Sneha",
        "Kiran", "Priya", "Rahul", "Divya", "Vijay",
        "Nisha", "Amit"
    ],
    "Department": [
        "CSE", "CSE", "ECE", "ECE", "CSE", "ECE",
        "CSE", "ECE", "CSE", "ECE", "CSE", "ECE"
    ],
    "Marks": [85, 72, 91, 65, 88, 74, 95, 58, 79, 82, 68, 90],
    "Attendance": [92, 78, 95, 68, 90, 80, 98, 60, 85, 88, 72, 94],
    "Study_Hours": [5, 3, 6, 2, 5, 3, 7, 2, 4, 5, 3, 6],
    "Assignment_Score": [90, 75, 94, 68, 89, 77, 96, 60, 82, 85, 70, 92]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.describe())
```

```python
plt.figure(figsize=(9, 4))
plt.bar(df["Student"], df["Marks"])
plt.title("Marks Obtained by Students")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
dept_avg = df.groupby("Department")["Marks"].mean()

print(dept_avg)

plt.figure(figsize=(6, 4))

plt.bar(dept_avg.index, dept_avg.values)

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.barh(df["Student"], df["Marks"])

plt.title("Student Marks Comparison")
plt.xlabel("Marks")
plt.ylabel("Student")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

bars = ax.bar(df["Student"], df["Marks"])

ax.bar_label(bars, fmt="%.0f", padding=3)
ax.set_title("Student Marks with Data Labels")
ax.set_ylabel("Marks")
ax.tick_params(axis="x", rotation=45)

# extra headroom so the top label isn't clipped
ax.set_ylim(0, df["Marks"].max() + 10)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

plt.hist(df["Marks"], bins=6, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap

rng = np.random.default_rng(42)

# Fake data: daily server load that slowly climbs, then blows up on day 45
days = np.arange(1, 61)
load = 30 + days * 0.9 + rng.normal(0, 3, days.size)
load[44:] = np.nan                      # nothing left after the explosion
boom_x, boom_y = 45, 30 + 45 * 0.9

fig, ax = plt.subplots(figsize=(11, 6))
fig.patch.set_facecolor("#0d0d14")
ax.set_facecolor("#0d0d14")
ax.set_xlim(0, 61)
ax.set_ylim(0, 110)

# The data line with a neon glow (several wide, faint copies underneath)
for lw, a in [(12, 0.05), (8, 0.08), (5, 0.15)]:
    ax.plot(days, load, color="#00e5ff", lw=lw, alpha=a, solid_capstyle="round")
ax.plot(days, load, color="#bff8ff", lw=2)

# Fireball: radial gradient rendered as an image, alpha fades to the edge
fire_cmap = LinearSegmentedColormap.from_list(
    "fire", ["#ffffff", "#fff3a0", "#ffb300", "#ff5a00", "#b3001b", "#2a0000"]
)
size = 400
yy, xx = np.mgrid[-1:1:size * 1j, -1:1:size * 1j]
r = np.hypot(xx, yy)
theta = np.arctan2(yy, xx)
# ragged edge: modulate the radius with a few random sine waves
ragged = r * (1 + 0.12 * np.sin(7 * theta + 1.3) + 0.08 * np.sin(13 * theta + 4.1)
              + 0.05 * np.sin(23 * theta))
rgba = fire_cmap(np.clip(ragged / 0.85, 0, 1))
# alpha hits 0 before the image border, so no square edges show
rgba[..., 3] = np.clip(1 - ragged / 0.85, 0, 1) ** 0.6

fx, fy = 19, 36                         # fireball half-size in data units
ax.imshow(rgba, extent=(boom_x - fx, boom_x + fx, boom_y - fy, boom_y + fy),
          origin="lower", zorder=5, aspect="auto", interpolation="bilinear")

# Smoke puffs around the fireball
for _ in range(40):
    ang = rng.uniform(0, 2 * np.pi)
    dist = rng.uniform(0.5, 1.0)
    ax.scatter(boom_x + np.cos(ang) * fx * dist, boom_y + np.sin(ang) * fy * dist,
               s=rng.uniform(800, 3000), color="#3a3a3a",
               alpha=rng.uniform(0.08, 0.2), lw=0, zorder=4)

# Shockwave rings
for k, a in zip([0.95, 1.1, 1.25], [0.5, 0.3, 0.15]):
    t = np.linspace(0, 2 * np.pi, 200)
    ax.plot(boom_x + np.cos(t) * fx * k, boom_y + np.sin(t) * fy * k,
            color="#ffd27a", lw=1.5, alpha=a, zorder=6)

# Rays bursting out
for ang in rng.uniform(0, 2 * np.pi, 28):
    length = rng.uniform(0.8, 1.3)
    ax.plot([boom_x + np.cos(ang) * fx * 0.4, boom_x + np.cos(ang) * fx * length],
            [boom_y + np.sin(ang) * fy * 0.4, boom_y + np.sin(ang) * fy * length],
            color="#ffcc33", lw=rng.uniform(0.5, 2), alpha=0.6, zorder=4)

# Sparks / debris
n = 250
ang = rng.uniform(0, 2 * np.pi, n)
dist = rng.exponential(0.35, n) + 0.2
ax.scatter(boom_x + np.cos(ang) * fx * dist, boom_y + np.sin(ang) * fy * dist,
           s=rng.uniform(2, 30, n), c=rng.uniform(0, 0.6, n), cmap=fire_cmap,
           zorder=7, lw=0)

# Comic-style label
ax.text(boom_x, boom_y + 2, "KABOOM!", ha="center", va="center", fontsize=40,
        fontweight="bold", color="#fff200", rotation=-8, zorder=10,
        path_effects=[pe.withStroke(linewidth=6, foreground="#c40000")])

ax.annotate("deployed on a Friday", xy=(boom_x - 3, boom_y - 4), xytext=(18, 95),
            color="white", fontsize=11, zorder=10,
            arrowprops=dict(arrowstyle="->", color="white", lw=1.2))

ax.set_title("Server Load Over Time", color="white", fontsize=16, pad=12)
ax.set_xlabel("Day", color="#aaaaaa")
ax.set_ylabel("Load (%)", color="#aaaaaa")
ax.tick_params(colors="#aaaaaa")
for spine in ax.spines.values():
    spine.set_color("#333344")
ax.grid(color="#222233", lw=0.6)

plt.tight_layout()
plt.show()
```
