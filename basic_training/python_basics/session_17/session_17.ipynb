```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 13500, 18000, 22000, 25000]
})
plt.figure(figsize=(8, 4))
plt.plot(
    sales_df["Month"],
    sales_df["Sales"],
    marker="o",
    linestyle="-",
    linewidth=2,
    label="Sales"
)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
sales_df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Product_A": [100, 120, 140, 135, 160],
    "Product_B": [90, 110, 115, 145, 155]
})

plt.figure(figsize=(8, 4))

plt.plot(sales_df["Month"], sales_df["Product_A"],
         marker="o", label="Product A")
plt.plot(sales_df["Month"], sales_df["Product_B"],
         marker="s", label="Product B")

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
data = {
    "Student": [
        "Anu", "Ravi", "Meera", "Arun", "Sneha",
        "Kiran", "Priya", "Rahul", "Divya", "Vijay",
        "Nisha", "Amit"
    ],
    "Department": [
        "CSE", "CSE", "ECE", "ECE", "CSE", "ECE",
        "CSE", "ECE", "CSE", "ECE", "CSE", "ECE"
    ],
    "Marks": [85, 72, 91, 65, 88, 74, 95, 58, 79, 82, 68, 90],
    "Attendance": [92, 78, 95, 68, 90, 80, 98, 60, 85, 88, 72, 94],
    "Study_Hours": [5, 3, 6, 2, 5, 3, 7, 2, 4, 5, 3, 6],
    "Assignment_Score": [90, 75, 94, 68, 89, 77, 96, 60, 82, 85, 70, 92]
}
df = pd.DataFrame(data)
print(df)
```

```python
sales = pd.DataFrame({
    "Region": [
        "South", "North", "South", "East", "North",
        "East", "South", "North", "East", "South"
    ],
    "Product": [
        "Laptop", "Mobile", "Mobile", "Laptop", "Laptop",
        "Mobile", "Laptop", "Mobile", "Laptop", "Mobile"
    ],
    "Sales": [
        50000, 30000, 25000, 45000, 55000,
        35000, 60000, 32000, 40000, 28000
    ],
    "Units_Sold": [5, 10, 8, 4, 6, 12, 7, 9, 4, 7],
    "Profit": [8000, 5000, 4000, 7000, 9000,
               6000, 10000, 4500, 6500, 4200]
})

print(sales)
```

```python
# Question 1: Which region generates the highest sales?
region_sales = sales.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(region_sales)
print("Top region:", region_sales.idxmax())

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(region_sales.index, region_sales.values)
ax.bar_label(bars, fmt="%.0f", padding=3)
ax.set_title("Total Sales by Region")
ax.set_xlabel("Region")
ax.set_ylabel("Sales")
ax.set_ylim(0, region_sales.max() * 1.1)
plt.tight_layout()
plt.show()
```

```python
# Question 2: Which product generates more profit?
product_profit = sales.groupby("Product")["Profit"].sum()
print(product_profit)
print("Most profitable product:", product_profit.idxmax())

plt.figure(figsize=(6, 4))
sns.barplot(data=sales, x="Product", y="Profit", estimator="sum", errorbar=None)
plt.title("Total Profit by Product")
plt.show()
```

```python
# Question 3: Is sales revenue related to profit?
plt.figure(figsize=(7, 4))
sns.regplot(data=sales, x="Sales", y="Profit")
plt.title("Sales vs Profit")
plt.show()
```

```python
# Question 4: What is the correlation between sales and profit?
sales_profit_corr = sales["Sales"].corr(sales["Profit"])
print("Correlation:", round(sales_profit_corr, 3))
```
