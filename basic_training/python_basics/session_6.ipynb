```python
names = ["Anu", "Ravi", "Meera"]
marks = [85,92,88]
result = {names[i]:marks[i] for i in range(len(names))}
print(result)
```

```python
names = ["Anu", "Ravi", "Meera"]
marks = [85,92,88]
result = {name:mark for name, mark in zip(names, marks)}
print(result)
```

```python
employees = {"Ravi":45000, "Meera":85000, "John":6200, "Anu":35000}
categories = {name:("High" if salary >=70000 else "Medium" if salary >= 50000 else "Low") for name, salary in employees.items()}
print(categories)
```

```python
transactions = [{"customer":"C101","amount":500}, {"customer":"C102","amount":800}, {"customer":"C101","amount":300}, {"customer":"C102","amount":900}]
totals = {}
for transaction in transactions:
    customer = transaction["customer"]
    amount = transaction["amount"]
    totals[customer] = totals.get(customer, 0)+amount
print(totals)
```

```python
transactions = [ 
    {"id": "T101", "amount": 500}, 
    {"id": "T102", "amount": 800}, 
    {"id": "T103", "amount": 300}, 
    {"id": "T102", "amount": 900}, 
    {"id": "T104", "amount": 700}, 
    {"id": "T101", "amount": 200}
]

transaction_ids = [t["id"] for t in transactions]
unique_ids = set(transaction_ids)
duplicate_ids = { transaction_id for transaction_id in
transaction_ids if transaction_ids.count(transaction_id) >
1}
print("Unique IDs:", unique_ids)
print("Duplicate IDs:", duplicate_ids)
```
