# AQI analyzer (NumPy)

```python
import numpy as np

data = np.loadtxt("aqi_data.csv", delimiter=",", skiprows=1, usecols=range(1, 31), dtype=int)
cities = ["Delhi", "Lucknow", "Kolkata", "Mumbai", "Hyderabad", "Chennai", "Bengaluru", "Mysuru"]
```

## SECTION 1: Know your data

```python
print("=== SECTION 1 ===")
print("1.1 Shape:", data.shape, "| ndim:", data.ndim, "| size:", data.size)
print("1.2 Delhi AQI on Day 15:", data[0, 14])
print("1.3 Chennai AQI last 7 days:", data[5, -7:])
print("1.4 First 5 days of every city:")
print(data[:, 0:5])
```

## SECTION 2: Statistics

```python
print("\n=== SECTION 2 ===")
means = np.mean(data, axis=1)
medians = np.median(data, axis=1)
stds = np.std(data, axis=1)
maxs = np.max(data, axis=1)
mins = np.min(data, axis=1)

print("2.1 Mean AQI per city:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} {means[i]:.2f}")

print("2.2 Median and standard deviation per city:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} median={medians[i]:.1f}  std={stds[i]:.2f}")

print("2.3 Max and min AQI per city:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} max={maxs[i]}  min={mins[i]}")

day_means = np.mean(data, axis=0)
worst_day = np.argmax(day_means) + 1
print("2.4 Day with highest average AQI:", worst_day,
      "| average:", round(np.max(day_means), 2))

p90 = np.percentile(data, 90)
print("2.5 90th percentile of all 240 readings:", p90)
```

## SECTION 3: Masking and np.where

```python
print("\n=== SECTION 3 ===")
days_above_200 = np.sum(data > 200, axis=1)
print("3.1 Days with AQI above 200 per city:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} {days_above_200[i]}")

print("3.2 City with most days above 200:", cities[np.argmax(days_above_200)])

delhi = data[0]
print("3.3 Delhi readings on Severe days (>400):", delhi[delhi > 400])

pct_above_100 = np.mean(data > 100, axis=1) * 100
print("3.4 Percentage of days above 100 per city:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} {pct_above_100[i]:.2f}%")

blr_labels = np.where(data[6] <= 100, "Safe", "Unsafe")
print("3.5 Bengaluru labels:", blr_labels)
print("    Safe days:", np.sum(blr_labels == "Safe"),
      "| Unsafe days:", np.sum(blr_labels == "Unsafe"))

excess = np.where(data > 100, data - 100, 0)
total_excess = np.sum(excess, axis=1)
print("3.6 Total excess AQI per city:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} {total_excess[i]}")
```

## SECTION 4: Trends and matrix operations

```python
print("\n=== SECTION 4 ===")
week1 = np.mean(data[:, 0:7], axis=1)
week4 = np.mean(data[:, 21:28], axis=1)
print("4.1 Week 1 vs Week 4 average AQI:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} week1={week1[i]:.2f}  week4={week4[i]:.2f}")

improvement = week1 - week4
trend = np.where(improvement > 0, "Improved", "Worsened")
print("4.2 Improvement (Week 1 - Week 4) and trend:")
for i in range(len(cities)):
    print(f"   {cities[i]:<10} {improvement[i]:>8.2f}  {trend[i]}")

data_t = data.T
print("4.3 Transposed shape:", data_t.shape)
print("    Day 1 row (all cities):", data_t[0])

load = data / np.max(data)
print("4.4 Delhi pollution load, first 5 days:", np.round(load[0, :5], 4))
```

## SECTION 5: Report

```python
print("\n=== SECTION 5: SUMMARY REPORT ===")
print(f"1) Most polluted city : {cities[np.argmax(means)]} (mean AQI {np.max(means):.2f})")
print(f"2) Cleanest city      : {cities[np.argmin(means)]} (mean AQI {np.min(means):.2f})")
print(f"3) Worst day          : Day {worst_day} (average AQI {np.max(day_means):.2f})")

max_pos = np.argmax(data)
num_days = data.shape[1]
max_city = cities[max_pos // num_days]
max_day = max_pos % num_days + 1
print(f"4) Highest reading    : {np.max(data)} ({max_city}, Day {max_day})")
print(f"5) Cities improved    : {np.sum(improvement > 0)} of {len(cities)}")

# Conclusion:
# Delhi and Lucknow need urgent action. Delhi was above 200 on 26 of 30 days, had 3 Severe days
# (up to 478) and got 71.29 points worse from Week 1 to Week 4. Lucknow was above 200 on 24 days
# and also got worse. Kolkata (mean 178.8) is worsening too and should be watched closely.
```
