```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", None, "Anil", "Meena", "dupe", "dupe"],
    "Age": [20, 21, None, 22, 1, 1],
    "Marks": [85, None, 67, 78, 1, 1]
})

print(df)

print(df.isnull())

print(df.isnull().sum())

print(df[df.isnull().any(axis=1)])
```

```python
clean = df.dropna()
print(clean)
```

```python
df["Marks"] = df["Marks"].fillna(0)
print(df)
```

```python
mean_marks = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(mean_marks)
print(df)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
print(df.duplicated())
```

```python
cleaned = df.drop_duplicates()
print(cleaned)
```

```python
import pandas as pd
data = {
    "Student_ID": [101, 102, 103, 104],
    "Name": ["  Rahul  ", " Priya", "Amit  ", "  Sneha  "],
    "Marks": [85, 90, 78, 88],
    "Gender": ["Male","F","male", "F"],
    "Date": [ "2026-01-15", "2026-02-20","2026-03-10", "2023-01-01"]
}
df = pd.DataFrame(data)
print(df)
df["Name"] = df["Name"].str.strip()
print(df)
```

```python
df["Name"] = df["Name"].str.upper()
print(df["Name"])
df["Name"] = df["Name"].str.lower()
print(df["Name"])
df["Name"] = df["Name"].str.title()
print(df["Name"])

df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "male": "Male"
})
print(df["Gender"])
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Year"] = df["Date"].dt.year
print(df["Year"])
df["Year"] = df["Date"].dt.month
print(df["Year"])
df["Year"] = df["Date"].dt.day
print(df["Year"])
df["Year"] = df["Date"].dt.day_name()
print(df["Year"])
```

```python
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Anil", "Meena", "Kiran"],
    "Department": ["CSE", "CSE", "ECE", "CSE", "ECE"],
    "Marks": [85, 92, 67, 78, 88]
})
result = df.groupby("Department")["Marks"].mean()
print(result)
```

```python
result = df.groupby("Department")["Marks"].sum()
print(result)
result = df.groupby("Department")["Marks"].max()
print(result)
result = df.groupby("Department")["Marks"].agg(
    ["mean", "max", "min", "count"]
)
print(result)
```
