  # Write DataFrame to Excel with a sheet name

```python
import pandas as pd

data = {
    "Name": ["Anil", "Priya", "Rahul"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_excel("students.xlsx", sheet_name="StudentDetails", index=False)
```

2. Use pd.ExcelWriter() when you want to write multiple DataFrames into the same Excel file.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Anil", "Priya"],
    "Marks": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Rahul", "Sneha"],
    "Marks": [78, 92]
})
 
with pd.ExcelWriter("college.xlsx") as writer:
    df2.to_excel(writer, sheet_name="Batch2", index=False)
    df1.to_excel(writer, sheet_name="Batch1", index=False)
```

```python
import pandas as pd

data = {
    "RollNo": [101, 102, 103],
    "Name": ["Anil", "Priya", "Rahul"],
    "Marks": [85, 92, 78]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)

df.to_json("students_output.json", orient="records", indent=4)
print("\nDataFrame successfully written to JSON file.")
```

```python
import pandas as pd
import json

students = [
    {"RollNo": 101, "Name": "Anil", "Marks": 85},
    {"RollNo": 102, "Name": "Priya", "Marks": 92},
    {"RollNo": 103, "Name": "Rahul", "Marks": 78}
]

with open("students.json", "w") as file:
    json.dump(students, file, indent=4)

df = pd.read_json("students.json")
print("Data loaded from JSON:")
print(df)
```

```python
df["Result"] =df["Marks"].apply(
    lambda mark: "Pass" if mark >= 40 else"Fail"
)

print("\nUpdated DataFrame:")
print(df)

df.to_json(
    "students_updated.json",
    orient="records",
    indent=4
)

print("\nUpdated data Saved to students_updated.json")
```

```python
import numpy as np
df["Result"] = np.where(
    df["Marks"] >= 40,
    "Pass",
    "Fail"
)
```

```python
def calculate_grade(mark):
    if mark >= 90:
        return "A+"
    elif mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 40:
        return "D"
    else:
        return "F"

df["Grade"] = df["Marks"].apply(calculate_grade)
```
