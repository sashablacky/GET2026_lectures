## Opening and closing files

````python
file = open("employees.txt", "r", encoding="utf-8")
content = file.read()
print(content)
file.close()
````

## The with open() statement

````python
with open("employees.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
````

````python
with open("student.txt", "w", encoding="utf-8") as file:
    file.write("Anil\n")
    file.write("Sunita\n")
    file.write("Rahul\n")
````

````python
with open("student.txt", "a", encoding="utf-8") as file:
    file.write("Meera\n")
````

````python
with open("new_report.txt", "x", encoding="utf-8") as file:
    file.write("Monthly report")
````

## Reading and writing CSV files
Suppose `employees.csv` contains:
```
id,name,salary
101,Anil,35000
102,Sunita,45000
103,Rahul,40000
```

### Reading a CSV file

````python
import csv

with open("employees.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
````

### Reading CSV using DictReader

````python
import csv

with open("employees.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["name"], row["salary"])
````

### Writing a CSV file using DictWriter

````python
import csv

employees = [
    {"id": 104, "name": "Meera", "salary": 50000},
    {"id": 105, "name": "Kiran", "salary": 42000}
]

with open("new_employees.csv", "w", newline="", encoding="utf-8") as file:
    fieldnames = ["id", "name", "salary"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(employees)
````

## Working with file positions: tell() and seek()

````python
with open("message.txt", "r", encoding="utf-8") as file:
    print(file.tell())

    print(file.read(7))
    print(file.tell())

    file.seek(0)
    print(file.read(5))
````

## JSON parsing: JSON to Python

````python
import json

json_data = '{"name": "Anil", "age": 25, "city": "Mangalore"}'

student = json.loads(json_data)

print(student)
print(type(student))
print(student["name"])
print(student["age"])
````

## Reading a JSON file using json.load()
Suppose `employee.json` contains:
```json
{
    "id": 101,
    "name": "Sunita",
    "department": "IT",
    "salary": 50000
}
```

````python
import json

with open("employee.json", "r", encoding="utf-8") as file:
    employee = json.load(file)

print(employee["name"])
print(employee["department"])
print(employee["salary"])
````

## JSON serialization: Python to JSON
### Convert a dictionary to a JSON string

````python
import json

employee = {
    "id": 101,
    "name": "Anil",
    "salary": 45000,
    "active": True
}

json_string = json.dumps(employee)
print(json_string)
print(type(json_string))
````

### For readable JSON, use indent

````python
import json

employee = {
    "id": 101,
    "name": "Anil",
    "skills": ["Python", "SQL"]
}

print(json.dumps(employee, indent=4))
````

### Writing a JSON file using json.dump()

````python
import json

employee = {
    "id": 101,
    "name": "Anil",
    "department": "IT",
    "salary": 45000,
    "skills": ["Python", "SQL"]
}

with open("employee.json", "w", encoding="utf-8") as file:
    json.dump(employee, file, indent=4)

print("JSON file created")
````

## Working with nested JSON

````python
import json

json_text = """
{
    "order_id": "ORD1001",
    "customer": {
        "name": "Anil",
        "city": "Mangalore"
    },
    "items": [
        {"product": "Laptop", "quantity": 1, "price": 55000},
        {"product": "Mouse", "quantity": 2, "price": 700}
    ]
}
"""

order = json.loads(json_text)
print(order["order_id"])
print(order["customer"]["name"])
print(order["customer"]["city"])
print(order["items"][0]["product"])
print(order["items"][1]["quantity"])
````

### Calculate the total order value

````python
total = 0

for item in order["items"]:
    total += item["quantity"] * item["price"]

print("Total order value:", total)
````

## Handling JSON arrays
### Find employees earning at least ₹45,000

````python
import json

json_text = """
[
    {"id": 101, "name": "Anil", "salary": 45000},
    {"id": 102, "name": "Sunita", "salary": 50000},
    {"id": 103, "name": "Rahul", "salary": 40000}
]
"""

employees = json.loads(json_text)

for employee in employees:
    if employee["salary"] >= 45000:
        print(employee["name"], employee["salary"])
````

## Converting JSON to Python and Python to JSON

````python
import json

# Python object
data = {
    "name": "Meera",
    "age": 28,
    "active": True,
    "skills": ["Python", "SQL"],
    "address": None
}

# Python -> JSON
json_text = json.dumps(data, indent=4)
print(json_text)

# JSON -> Python
restored_data = json.loads(json_text)
print(restored_data)
print(type(restored_data))
````
