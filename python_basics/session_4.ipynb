```python
elements = [10, 20, 30, 40, 50]
while True:
    choice = input("Enter a choice: \n"
    "1. Append an element\n"
    "2. Insert an element at a specific index\n"
    "3. Pop an element at a specific index\n"
    "4. Find an element\n"
    "5. Get the length of the list\n"
    "6. Get the maximum element in the list\n"
    "7. Get the minimum element in the list\n"
    "8. Sort the list in ascending order\n"
    "9. Sort the list in descending order\n"
    "10. Reverse the list\n"
    "11. Display the current list\n")
    if choice == '12':
        break
    if choice == '1':
        elements.append(int(input("Enter an element to append: ")))
    elif choice == '2':
        index = int(input("Enter an index to insert at: "))
        elements.insert(index, int(input("Enter an element to insert: ")))
    elif choice == '3':
        index = int(input("Enter an index to pop: "))
        elements.pop(index)
    elif choice == '4':
        element = int(input("Enter an element to find: "))
        if element in elements:
            print(f"{element} is at index {elements.index(element)}")
        else:
            print(f"{element} is not in the list.")
    elif choice == '5':
        print("Current amount of elements in the list:" , len(elements))
    elif choice == '6':
        print("Current maximum element in the list:" , max(elements))
    elif choice == '7':
        print("Current minimum element in the list:" , min(elements))
    elif choice == '8':
        print("List of elements after sorting:" , elements.sort())
    elif choice == '9':
        print("List of elements in reverse order:" , elements.sort(reverse=True))
    elif choice == '10':
        print("reversed list of elements:" , list(reversed(elements)))
    elif choice == '11':
        print("Current list of elements:" , elements)
```
