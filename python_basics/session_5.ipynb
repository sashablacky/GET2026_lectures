```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))
print(a | b)
print(a.intersection(b))
print(a & b)
print(a ^ b)
```

```python
students = ["Rahul", "Priya", "Anil", "Rahul", "Meena", "Priya"]
#1
print("Students:", students)
#2
print("Total students:", len(students))
#3
students.append("Kiran")
#4
students.remove("Anil")
#5
seen = []
duplicates = []
for student in students:
    if student in seen:
        if student not in duplicates:
            duplicates.append(student)
    else:
        seen.append(student)

print("Duplicates:", duplicates)
#6
students_set = set(students)
print("Unique students:", students_set)
#7
print("Sorted unique students:", sorted(students_set))
#8
for student in students_set:
    print(student.upper())
#9
students_string = ", ".join(students)
print("Students:", students_string)
```
