```
m1,m2,m3,m4,m5 = map(int, input("input here").split())

total = m1 + m2 + m3 + m4 + m5
average = total / 5
passed = m1>=40 and m2>=40 and m3>=40 and m4>=40 and m5>=40
percentage = (total / 5)*100

def assign_letter(mark):
    if mark >= 90:
        return "A+"
    elif mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 50:
        return "D"
    else:
        return "F"

print(f"total marks {total}\n"
      f"average {average}\n"
      f"passed {passed}\n"
      f"percentage {percentage}%\n"
      f"Mark 1 {assign_letter(m1)}\n"
      f"Mark 2 {assign_letter(m2)}\n"
      f"Mark 3 {assign_letter(m3)}\n"
      f"Mark 4 {assign_letter(m4)}\n"
      f"Mark 5 {assign_letter(m5)}\n")
```

```
basic_salary = float(input("Enter basic salary: "))

DA = 0.20
HRA = 0.15
PF = 0.12

gross_salary = basic_salary + basic_salary * DA + basic_salary * HRA

if basic_salary >= 50000:
    gross_salary += basic_salary * 0.05

pf_amount = basic_salary * PF
net_salary = gross_salary - pf_amount

print(f"Gross salary: {gross_salary}")
print(f"Net salary: {net_salary}")
```

```
age = int(input("Enter age: "))
income = int(input("Enter monthly income: "))
credit_score = int(input("Enter credit score: "))

if 20 <= age <= 60 and income >= 25000 and credit_score >= 700:
    P = float(input("Loan amount: "))
    annual_rate = float(input("Annual interest rate (%): "))
    n = int(input("Number of monthly payments: "))

    r = annual_rate / 12 / 100
    EMI = (P * r * (1 + r) ** n) / ((1 + r) ** n - 1)
    print(f"Eligible. EMI: {EMI:.2f}")
else:
    print("Not eligible")
```

```
class BankAccount:
    def __init__(self, name, balance, pin):
        self.name = name
        self.balance = balance
        self.pin = pin

    def withdraw(self, amount):
        if amount > 0 and amount % 100 == 0 and amount <= 20000 and amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawal of {amount} from {self.name}'s account successful")
            print(f"Remaining balance: {self.balance}")
        else:
            print(f"Withdrawal of {amount} from {self.name}'s account failed")


account1 = BankAccount("Alex", 100000, "0000")

pin = input("Enter PIN: ")

if pin == account1.pin:
    withdraw_amount = int(input("Enter withdraw amount: "))
    account1.withdraw(withdraw_amount)
else:
    print("Wrong PIN")
```
